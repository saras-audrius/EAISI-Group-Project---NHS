```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 📁 List of CSV files
csv_bestanden = ["C:\\Users\\olifa\\OneDrive\\Documenten\\GitHub\\EAISI-Group-Project---NHS\\data\\interim\\hip-ccg.csv",
                 "C:\\Users\\olifa\\OneDrive\\Documenten\\GitHub\\EAISI-Group-Project---NHS\\data\\interim\\hip-provider.csv",
                 "C:\\Users\\olifa\\OneDrive\\Documenten\\GitHub\\EAISI-Group-Project---NHS\\data\\interim\\knee-ccg.csv",
                 "C:\\Users\\olifa\\OneDrive\\Documenten\\GitHub\\EAISI-Group-Project---NHS\\data\\interim\\knee-provider.csv"]

# 🔁 Go through each file
for pad in csv_bestanden:
    if not os.path.exists(pad):
        print(f"⚠️ File not found: {pad}")
        continue

    try:
        df = pd.read_csv(pad)

        # 🔍 Missende waarden heatmap
        plt.figure(figsize=(12, 6))
        sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
        plt.title(f"Missende waarden in {pad}", fontsize=14)
        plt.xlabel("Kolommen")
        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Fout bij verwerken van {pad}: {e}")
```

The 20 highest coeffients per file

```python
import numpy as np

import glob
csv_bestanden = glob.glob("C:\\Users\\olifa\\OneDrive\\Documenten\\GitHub\\EAISI-Group-Project---NHS\\data\\interim\\*.csv")

for pad in csv_bestanden:
    print(f"\n📄 Bestand: {pad}")
    
    if not os.path.exists(pad):
        print("⚠️ Bestand niet gevonden.")
        continue

    try:
        df = pd.read_csv(pad)
        numeriek_df = df.select_dtypes(include=[np.number])

        if numeriek_df.shape[1] < 2:
            print("📌 Onvoldoende numerieke kolommen voor correlatieanalyse.")
            continue

        # 🔢 Calculate correlation matrix
        corr_matrix = numeriek_df.corr()

        # 🔥 Heatmap of correlations
        plt.figure(figsize=(max(10, len(corr_matrix.columns)), 8))
        sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", square=True)
        plt.title(f"Correlatiematrix voor {pad}", fontsize=14)
        plt.tight_layout()
        plt.show()

        # 🔍 Top 20 strongest correlations
        corr_long = corr_matrix.stack().reset_index()
        corr_long.columns = ['Column1', 'Column2', 'Correlation']
        corr_long = corr_long[corr_long['Column1'] != corr_long['Column2']]
        corr_long['pair'] = corr_long.apply(lambda x: tuple(sorted([x['Column1'], x['Column2']])), axis=1)
        corr_long = corr_long.drop_duplicates(subset='pair')
        corr_long['AbsCorrelation'] = corr_long['Correlation'].abs()
        top20 = corr_long.sort_values(by='AbsCorrelation', ascending=False).head(20)

        print("\n🔝 Top 20 strongest correlations:")
        for i, row in top20.iterrows():
            print(f"{row['Column1']} ↔ {row['Column2']}: {row['Correlation']:.4f}")

    except Exception as e:
        print(f"❌ Error processing {pad}: {e}")
```
